```python
source_path = "/Volumes/sahil_catalog/taxi/taxi/"

df_raw =( spark.read
            .format("parquet")
            .option("header", "true")
            .load(source_path)
)
```

```python
df_raw.count()
df_raw.printSchema()
```

```python
from pyspark.sql.functions import current_timestamp, col

df_bronze = (
    df_raw
    .withColumn("inserted_at", current_timestamp())
    .withColumn("source_system", col("_metadata.file_path"))
)

(
    df_bronze.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("sahil_catalog.taxi.bronze_nyc_taxi")
)
```

```python
%sql
SELECT * FROM sahil_catalog.taxi.bronze_nyc_taxi LIMIT 10;
```
